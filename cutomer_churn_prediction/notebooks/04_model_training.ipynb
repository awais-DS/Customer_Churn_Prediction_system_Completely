```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
```

```python
df=pd.read_csv("https://raw.githubusercontent.com/awais-DS/Data/refs/heads/main/clean_data.csv")
```

```python
x=df.drop(columns=["Churn"])
y=df["Churn"]
```

```python
x.shape,y.shape
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=0,stratify=y)
```

```python
# Creating an emty list to collect all algoritms metrics
performance_metrics=[]
```

I am going to train five classification model and will select te best one.

```python
# 1 Logistics Regression
# Importing from Sklearn
from sklearn.linear_model import LogisticRegression

# intializing the model
lr=LogisticRegression(class_weight="balanced",random_state=0,max_iter=1000)

# fitting/traing on training set
lr.fit(x_train,y_train)

# predicting on test data
lr_predict=lr.predict(x_test)

#Evaluating with actual values
lr_accuracy=accuracy_score(y_test,lr_predict)
lr_precision=precision_score(y_test,lr_predict)
lr_recall=recall_score(y_test,lr_predict)
lr_f1=f1_score(y_test,lr_predict)


#printing the scores
print("Logistics Regression Metrics Score")
print(f"Accuracy : {lr_accuracy:.4f}")
print(f"Precision : {lr_precision:.4f}")
print(f"Recall : {lr_recall:.4f}")
print(f"F1 Score: {lr_f1:.4f}")

# Saving scores for final table
performance_metrics.append({
    "Algorithm":"Logistic Regression",
    "Accuracy": lr_accuracy,
    "Precision": lr_precision,
    "Recall":lr_recall,
    "F1 Score":lr_f1,
}) 
```

```python
# 2 DecisionTreeClassifier
# importing and initializing
from sklearn.tree import DecisionTreeClassifier
dt=DecisionTreeClassifier(random_state=0,class_weight="balanced",max_depth=6)

#fitting
dt.fit(x_train,y_train)

#predicting
dt_predict=dt.predict(x_test)

# Evalusting
dt_accuracy=accuracy_score(y_test,dt_predict)
dt_precision=precision_score(y_test,dt_predict)
dt_recall=recall_score(y_test,dt_predict)
dt_f1=f1_score(y_test,dt_predict)

# printing the metrics
print("Decision Tree Classifier Metrics")
print(f"Accuracy: {dt_accuracy:.4f}")
print(f"Precision : {dt_precision:.4f}")
print(f"Recall:{dt_recall:.4f}")
print(f"F1 Score:{dt_f1:.4f}")


#Saving all scores to Metrics performance
performance_metrics.append({
    "Algorithm":"Decision Tree",
    "Accuracy":dt_accuracy,
    "Precision":dt_precision,
    "Recall":dt_recall,
    "F1-Score":dt_f1,

})
```

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(
    n_estimators=300,
    class_weight="balanced",
    min_samples_leaf=5,
    random_state=0,
    n_jobs=-1,
)
rf.fit(x_train, y_train)
rf_predict = rf.predict(x_test)

rf_accuracy  = accuracy_score(y_test, rf_predict)
rf_precision = precision_score(y_test, rf_predict)
rf_recall    = recall_score(y_test, rf_predict)
rf_f1        = f1_score(y_test, rf_predict)

print("Random Forest Classifier Metrics")
print(f"Accuracy : {rf_accuracy:.4f}")
print(f"Precision : {rf_precision:.4f}")
print(f"Recall : {rf_recall:.4f}")
print(f"F1 Score : {rf_f1:.4f}")

performance_metrics.append({
    "Algorithm": "Random Forest",
    "Accuracy": rf_accuracy,
    "Precision": rf_precision,
    "Recall": rf_recall,
    "F1 Score": rf_f1,
})
```

```python
# 3 Support Vector Classifier
from sklearn.svm import SVC

# initializing the model
svc = SVC(kernel="rbf", class_weight="balanced", random_state=0)

# fitting/training on training set
svc.fit(x_train, y_train)

# predicting on test data
svc_predict = svc.predict(x_test)

# evaluating with actual values
svc_accuracy  = accuracy_score(y_test, svc_predict)
svc_precision = precision_score(y_test, svc_predict)
svc_recall    = recall_score(y_test, svc_predict)
svc_f1        = f1_score(y_test, svc_predict)

# printing the scores
print("Support Vector Classifier Metrics")
print(f"Accuracy : {svc_accuracy:.4f}")
print(f"Precision : {svc_precision:.4f}")
print(f"Recall : {svc_recall:.4f}")
print(f"F1 Score : {svc_f1:.4f}")

# overfitting check
svc_train_acc = accuracy_score(y_train, svc.predict(x_train))
print(f"Train Accuracy: {svc_train_acc:.4f}")
print(f"Test Accuracy:  {svc_accuracy:.4f}")

# saving scores for final table
performance_metrics.append({
    "Algorithm": "SVC",
    "Accuracy": svc_accuracy,
    "Precision": svc_precision,
    "Recall": svc_recall,
    "F1 Score": svc_f1,
})
```

```python
# 5 XGBoost Classifier
from xgboost import XGBClassifier



# initializing the model
xgb = XGBClassifier(
    n_estimators=300,
    learning_rate=0.05,
    max_depth=4,
    subsample=0.8,
    colsample_bytree=0.8,
    scale_pos_weight=spw,
    eval_metric="logloss",
    random_state=0,
    n_jobs=-1,
)

# fitting/training on training set
xgb.fit(x_train, y_train)

# predicting on test data
xgb_predict = xgb.predict(x_test)

# evaluating with actual values
xgb_accuracy  = accuracy_score(y_test, xgb_predict)
xgb_precision = precision_score(y_test, xgb_predict)
xgb_recall    = recall_score(y_test, xgb_predict)
xgb_f1        = f1_score(y_test, xgb_predict)

# printing the scores
print("XGBoost Classifier Metrics")
print(f"Accuracy : {xgb_accuracy:.4f}")
print(f"Precision : {xgb_precision:.4f}")
print(f"Recall : {xgb_recall:.4f}")
print(f"F1 Score : {xgb_f1:.4f}")

# overfitting check
xgb_train_acc = accuracy_score(y_train, xgb.predict(x_train))
print(f"Train Accuracy: {xgb_train_acc:.4f}")
print(f"Test Accuracy:  {xgb_accuracy:.4f}")

# saving scores for final table
performance_metrics.append({
    "Algorithm": "XGBoost",
    "Accuracy": xgb_accuracy,
    "Precision": xgb_precision,
    "Recall": xgb_recall,
    "F1 Score": xgb_f1,
})
```
